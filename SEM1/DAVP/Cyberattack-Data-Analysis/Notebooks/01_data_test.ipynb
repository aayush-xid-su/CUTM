```python
import numpy as np

# Select numerical columns
numerical_df = df.select_dtypes(include='number')

# Count positive and negative infinity values
positive_inf = np.isposinf(numerical_df).sum().sum()
negative_inf = np.isneginf(numerical_df).sum().sum()

print("Positive infinity values:", positive_inf)
print("Negative infinity values:", negative_inf)
print("Total infinity values:", positive_inf + negative_inf)
```

```python
# Count infinity values in each numerical column
inf_counts = pd.Series(
    np.isinf(numerical_df).sum(),
    index=numerical_df.columns
)

# Display only columns containing infinity
print(inf_counts[inf_counts > 0])
```

```python
# Count missing values in each column
missing_counts = df.isnull().sum()

print("Total missing values:", missing_counts.sum())

print("\nColumns containing missing values:")
print(missing_counts[missing_counts > 0])
```

```python
# Check the exact names of the affected columns
for col in df.columns:
    if 'Label' in col or 'Flow Duration' in col:
        print(repr(col))
```

```python
# Find the label column, ignoring leading/trailing spaces
label_col = next(
    col for col in df.columns
    if col.strip() == 'Label'
)

print("Actual label column:", repr(label_col))
```

```python
# Identify rows containing infinity
inf_mask = np.isinf(
    df[['Flow Bytes/s', 'Flow Packets/s']]
).any(axis=1)

print("Rows containing infinity:", inf_mask.sum())

# Display affected records
affected_rows = df.loc[
    inf_mask,
    ['Flow Duration', 'Flow Bytes/s',
     'Flow Packets/s', 'Label']
]

display(affected_rows.head(10))
```

```python
# Identify rows with missing Flow Bytes/s
missing_mask = df['Flow Bytes/s'].isna()

print("Rows with missing Flow Bytes/s:", missing_mask.sum())

# Display affected records
missing_rows = df.loc[
    missing_mask,
    ['Flow Duration',
     'Total Length of Fwd Packets',
     'Total Length of Bwd Packets',
     'Flow Bytes/s',
     'Label']
]

display(missing_rows.head(10))
```
