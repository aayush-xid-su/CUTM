```python

import pandas as pd
import numpy as np

file_path = "../Dataset/Wednesday-workingHours.pcap_ISCX.csv"
df = pd.read_csv(file_path)


df_clean = df.copy()

# Display original dataset dimensions
print("Original Dataset Shape:", df.shape)

# Display working dataset dimensions
print("Working Dataset Shape:", df_clean.shape)

# Verify that both DataFrames initially contain the same data
print("\nOriginal and working datasets are identical:",
      df.equals(df_clean))

# Verify that they are separate DataFrame objects
print("Are they separate objects?:", df is not df_clean)

print("\nWorking DataFrame created successfully!")
```

```python
# Store the original column names
original_columns = df_clean.columns.tolist()

# Remove leading and trailing spaces
cleaned_columns = [column.strip() for column in original_columns]

# Check whether removing spaces creates duplicate column names
duplicate_mask = pd.Index(cleaned_columns).duplicated(keep=False)

if duplicate_mask.any():

    print("WARNING: Duplicate column names found after removing spaces!")

    duplicate_report = pd.DataFrame({
        "Original Column Name": original_columns,
        "Cleaned Column Name": cleaned_columns
    })

    print(duplicate_report[duplicate_mask])

else:

    # Apply standardized column names
    df_clean.columns = cleaned_columns

    print("Column names standardized successfully!")

# Display the first 15 column names
print("\nFirst 15 column names:")
print(df_clean.columns[:15].tolist())

# Check for remaining leading or trailing spaces
remaining_spaces = [
    column for column in df_clean.columns
    if column != column.strip()
]

print("\nColumn names with unwanted spaces:", remaining_spaces)

# Display the total number of columns
print("\nTotal columns:", len(df_clean.columns))
```

```python
# Count missing values in each column
missing_count = df_clean.isnull().sum()

# Calculate missing-value percentage
missing_percentage = (missing_count / len(df_clean)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

# Display only columns containing missing values
missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
]

# Sort by missing count in descending order
missing_summary = missing_summary.sort_values(
    by="Missing Count",
    ascending=False
)

print("Missing Values Summary:")
print(missing_summary)

# Calculate total missing cells
total_missing = df_clean.isnull().sum().sum()

# Count rows containing at least one missing value
rows_with_missing = df_clean.isnull().any(axis=1).sum()

print("\nTotal Missing Values:", total_missing)
print("Rows Containing Missing Values:", rows_with_missing)

# Calculate the percentage of rows affected
percentage_rows = (rows_with_missing / len(df_clean)) * 100

print("Percentage of Affected Rows:", round(percentage_rows, 4), "%")
```

```python
# Select numerical columns
numerical_data = df_clean.select_dtypes(include=[np.number])

# Identify positive and negative infinity
positive_inf = np.isposinf(numerical_data)
negative_inf = np.isneginf(numerical_data)

# Count infinity values
positive_inf_count = positive_inf.sum().sum()
negative_inf_count = negative_inf.sum().sum()

total_inf = positive_inf_count + negative_inf_count

print("Positive Infinity Values:", positive_inf_count)
print("Negative Infinity Values:", negative_inf_count)
print("Total Infinity Values:", total_inf)

# Count infinity values in each column
inf_summary = pd.DataFrame({
    "Positive Infinity": positive_inf.sum(),
    "Negative Infinity": negative_inf.sum()
})

# Display only columns containing infinity
inf_summary = inf_summary[
    (inf_summary["Positive Infinity"] > 0) |
    (inf_summary["Negative Infinity"] > 0)
]

print("\nInfinity Values by Column:")
print(inf_summary)

# Count rows containing at least one infinity value
rows_with_inf = (positive_inf | negative_inf).any(axis=1).sum()

print("\nRows Containing Infinity:", rows_with_inf)
```

```python
# Count completely duplicated rows
duplicate_count = df_clean.duplicated().sum()

# Calculate duplicate percentage
duplicate_percentage = (
    duplicate_count / len(df_clean)
) * 100

print("Total Duplicate Rows:", duplicate_count)

print(
    "Duplicate Percentage:",
    round(duplicate_percentage, 4),
    "%"
)

# Display a few duplicate records if they exist
if duplicate_count > 0:

    print("\nSample Duplicate Records:")

    duplicate_samples = df_clean[
        df_clean.duplicated(keep=False)
    ].head(5)

    display(duplicate_samples)

else:

    print("\nNo duplicate records found.")
```

```python
# Features that should not contain negative values
non_negative_features = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Fwd Packet Length Max",
    "Fwd Packet Length Min",
    "Bwd Packet Length Max",
    "Bwd Packet Length Min",
    "Flow Bytes/s",
    "Flow Packets/s"
]

# Keep only features that actually exist
available_features = [
    col for col in non_negative_features
    if col in df_clean.columns
]

# Identify negative values
negative_summary = {}

for col in available_features:
    negative_count = (df_clean[col] < 0).sum()
    negative_summary[col] = negative_count

# Convert results into a DataFrame
negative_report = pd.DataFrame(
    list(negative_summary.items()),
    columns=["Feature", "Negative Value Count"]
)

# Display all selected features
print("Negative Value Summary:")
display(negative_report)

# Display features containing negative values
invalid_features = negative_report[
    negative_report["Negative Value Count"] > 0
]

print("\nFeatures Containing Negative Values:")
if len(invalid_features) > 0:
    display(invalid_features)
else:
    print("No negative values found in the selected features.")
```

```python
# Features containing negative values
negative_features = [
    "Flow Duration",
    "Flow Bytes/s",
    "Flow Packets/s"
]

# Create a mask for rows containing negative values
negative_mask = (
    df_clean[negative_features] < 0
).any(axis=1)

# Select useful columns for investigation
inspection_columns = [
    "Flow ID",
    "Timestamp",
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Label"
]

# Keep only columns available in the dataset
available_columns = [
    col for col in inspection_columns
    if col in df_clean.columns
]

# Display affected records
negative_records = df_clean.loc[
    negative_mask, available_columns
]

print("Total Records With Negative Values:",
      len(negative_records))

display(negative_records)

# Count negative values separately in affected records
print("\nNegative Values by Feature:")
for col in negative_features:
    print(col, ":", (df_clean[col] < 0).sum())
```

```python
# Identify records with zero flow duration
zero_duration_mask = df_clean["Flow Duration"] == 0

# Count zero-duration records
zero_duration_count = zero_duration_mask.sum()

print("Records With Zero Flow Duration:",
      zero_duration_count)

# Check missing and infinite rates in zero-duration records
zero_duration_records = df_clean.loc[zero_duration_mask]

print("\nMissing Flow Bytes/s in Zero-Duration Records:",
      zero_duration_records["Flow Bytes/s"].isna().sum())

print("Infinite Flow Bytes/s in Zero-Duration Records:",
      np.isinf(zero_duration_records["Flow Bytes/s"]).sum())

print("Infinite Flow Packets/s in Zero-Duration Records:",
      np.isinf(zero_duration_records["Flow Packets/s"]).sum())

# Display a few examples
inspection_columns = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Label"
]

print("\nSample Zero-Duration Records:")
display(zero_duration_records[inspection_columns].head(10))
```

```python
# Display timestamp data type
print("Timestamp Data Type:",
      df_clean["Timestamp"].dtype)

# Display first 10 timestamp values
print("\nFirst 10 Timestamp Values:")
display(df_clean["Timestamp"].head(10))

# Display last 10 timestamp values
print("\nLast 10 Timestamp Values:")
display(df_clean["Timestamp"].tail(10))

# Check missing timestamps
print("\nMissing Timestamp Values:",
      df_clean["Timestamp"].isna().sum())

# Display unique timestamp examples
print("\nSample Unique Timestamps:")
display(df_clean["Timestamp"].dropna().unique()[:10])
```

```python
# Convert timestamp strings into datetime values
timestamp_converted = pd.to_datetime(
    df_clean["Timestamp"],
    format="%d/%m/%Y %H:%M",
    errors="coerce"
)

# Check whether any values failed conversion
conversion_errors = timestamp_converted.isna().sum()

print("Timestamp Conversion Errors:", conversion_errors)

# Apply conversion only if all values are valid
if conversion_errors == 0:

    df_clean["Timestamp"] = timestamp_converted

    print("\nTimestamp converted successfully!")

else:

    print("\nSome timestamps could not be converted.")
    print("Please inspect the affected records.")

# Display updated data type
print("\nUpdated Timestamp Data Type:",
      df_clean["Timestamp"].dtype)

# Display sample converted timestamps
print("\nFirst 10 Converted Timestamps:")
display(df_clean["Timestamp"].head(10))

# Display earliest and latest timestamps
print("\nEarliest Timestamp:",
      df_clean["Timestamp"].min())

print("Latest Timestamp:",
      df_clean["Timestamp"].max())
```

```python
# Features requiring cleaning
rate_features = [
    "Flow Bytes/s",
    "Flow Packets/s"
]

# ------------------------------------------
# 1. Replace infinite values with NaN
# ------------------------------------------

for col in rate_features:

    df_clean[col] = df_clean[col].replace(
        [np.inf, -np.inf],
        np.nan
    )

print("Infinite values replaced with NaN.")

# ------------------------------------------
# 2. Replace negative flow duration
# ------------------------------------------

negative_duration_count = (
    df_clean["Flow Duration"] < 0
).sum()

df_clean.loc[
    df_clean["Flow Duration"] < 0,
    "Flow Duration"
] = np.nan

print(
    "\nNegative Flow Duration values replaced:",
    negative_duration_count
)

# ------------------------------------------
# 3. Replace negative traffic rates
# ------------------------------------------

for col in rate_features:

    negative_count = (df_clean[col] < 0).sum()

    df_clean.loc[
        df_clean[col] < 0,
        col
    ] = np.nan

    print(
        f"Negative values replaced in {col}:",
        negative_count
    )

# ------------------------------------------
# 4. Verify the cleaning
# ------------------------------------------

print("\nRemaining Invalid Values:")

for col in ["Flow Duration"] + rate_features:

    negative_remaining = (df_clean[col] < 0).sum()

    infinite_remaining = np.isinf(df_clean[col]).sum()

    missing_count = df_clean[col].isna().sum()

    print(f"\n{col}")
    print("Negative values:", negative_remaining)
    print("Infinite values:", infinite_remaining)
    print("Missing values:", missing_count)

# ------------------------------------------
# 5. Verify dataset dimensions
# ------------------------------------------

print("\nDataset Shape After Cleaning:", df_clean.shape)
```

```python
# Count missing values in every column
missing_count = df_clean.isna().sum()

# Calculate missing percentage
missing_percentage = (
    missing_count / len(df_clean)
) * 100

# Create missing-value summary
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

# Keep only columns containing missing values
missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
]

# Sort by missing count
missing_summary = missing_summary.sort_values(
    by="Missing Count",
    ascending=False
)

print("Missing Values After Cleaning:")
display(missing_summary)

# Calculate total missing cells
total_missing = df_clean.isna().sum().sum()

# Calculate affected rows
affected_rows = df_clean.isna().any(axis=1).sum()

# Calculate affected row percentage
affected_percentage = (
    affected_rows / len(df_clean)
) * 100

print("\nTotal Missing Values:", total_missing)
print("Rows Containing Missing Values:", affected_rows)

print(
    "Percentage of Affected Rows:",
    round(affected_percentage, 4),
    "%"
)
```

```python
# Select the three affected features
missing_features = [
    "Flow Duration",
    "Flow Bytes/s",
    "Flow Packets/s"
]

# Identify rows containing missing values
missing_mask = df_clean[missing_features].isna().any(axis=1)

affected_data = df_clean.loc[missing_mask]

# ------------------------------------------
# 1. Count missing values by feature
# ------------------------------------------

print("Missing Values by Feature:")
print(affected_data[missing_features].isna().sum())

# ------------------------------------------
# 2. Examine combinations of missing values
# ------------------------------------------

print("\nMissing-Value Combinations:")

missing_patterns = (
    affected_data[missing_features]
    .isna()
    .value_counts()
    .reset_index(name="Record Count")
)

display(missing_patterns)

# ------------------------------------------
# 3. Examine affected traffic labels
# ------------------------------------------

print("\nTraffic Labels Among Affected Records:")

label_distribution = affected_data["Label"].value_counts()

display(label_distribution)

# ------------------------------------------
# 4. Calculate affected records percentage
#    for each traffic label
# ------------------------------------------

print("\nAffected Records by Traffic Label:")

label_report = pd.DataFrame({
    "Affected Records": affected_data["Label"].value_counts(),
    "Total Records": df_clean["Label"].value_counts()
})

label_report["Affected Percentage"] = (
    label_report["Affected Records"] /
    label_report["Total Records"]
) * 100

display(label_report.fillna(0).round(4))
```

```python
# Select numerical columns
numeric_columns = df_clean.select_dtypes(include=np.number).columns

# Count remaining infinite values
infinity_count = np.isinf(df_clean[numeric_columns]).sum().sum()

# Count remaining negative values in selected features
features_to_check = [
    "Flow Duration",
    "Flow Bytes/s",
    "Flow Packets/s"
]

print("Remaining Infinite Values:", infinity_count)

print("\nRemaining Negative Values:")
for column in features_to_check:
    negative_count = (df_clean[column] < 0).sum()
    print(f"{column}: {negative_count}")

# Check missing values in the entire dataset
print("\nTotal Missing Values:", df_clean.isna().sum().sum())

# Verify dataset dimensions
print("\nFinal Dataset Shape:", df_clean.shape)
```

```python
# Identify duplicate rows
duplicate_mask = df_clean.duplicated(keep=False)

# Store all duplicate records
duplicate_records = df_clean.loc[duplicate_mask]

print("Total Duplicate Records:", duplicate_records.shape[0])

# Examine traffic labels among duplicate records
print("\nDuplicate Records by Traffic Label:")
display(duplicate_records["Label"].value_counts())

# Examine duplicate groups
print("\nNumber of Duplicate Groups:")
print(df_clean.duplicated().sum())

# Display a sample of duplicate records
print("\nSample Duplicate Records:")
display(
    duplicate_records[
        ["Flow ID", "Source IP", "Destination IP",
         "Timestamp", "Label"]
    ].head(10)
)
```

```python
# Load the saved cleaned dataset
df_verify = pd.read_csv(
    "../Dataset/Wednesday_Cleaned.csv.gz",
    compression="gzip",
    parse_dates=["Timestamp"]
)

# Compare dataset dimensions
print("Original Cleaned DataFrame Shape:", df_clean.shape)
print("Reloaded Dataset Shape:", df_verify.shape)

# Verify column names
print("\nColumn Names Match:",
      df_clean.columns.equals(df_verify.columns))

# Verify missing values
print("Missing Values Match:",
      df_clean.isna().sum().sum() == df_verify.isna().sum().sum())

# Verify traffic label distribution
print("Label Distribution Match:",
      df_clean["Label"].value_counts().sort_index().equals(
          df_verify["Label"].value_counts().sort_index()
      ))

# Final verification
if (
    df_clean.shape == df_verify.shape
    and df_clean.columns.equals(df_verify.columns)
    and df_clean.isna().sum().sum() == df_verify.isna().sum().sum()
    and df_clean["Label"].value_counts().sort_index().equals(
        df_verify["Label"].value_counts().sort_index()
    )
):
    print("\nSUCCESS: Saved dataset verified!")
else:
    print("\nWARNING: Dataset verification needs investigation.")
```
