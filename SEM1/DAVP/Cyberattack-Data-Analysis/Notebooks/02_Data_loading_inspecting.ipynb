```python
# Import Pandas for data manipulation and analysis
import pandas as pd

# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for data visualization
import matplotlib.pyplot as plt

# Import Seaborn for statistical visualization
import seaborn as sns
```

```python
# Define the dataset file path
file_path = "../Dataset/Wednesday-workingHours.pcap_ISCX.csv"

# Load the CSV dataset into a Pandas DataFrame
df = pd.read_csv(file_path)

# Display a confirmation message
print("Dataset loaded successfully!")
```

```python
# Display the data type of every column
print("Column Data Types:\n")

print(df.dtypes)
```

```python
# Find the target column without assuming its exact spacing
label_columns = [
    col for col in df.columns
    if col.strip() == "Label"
]

print("Target column found:", label_columns)
```

```python
# Identify the exact target column name
label_col = label_columns[0]

# Count records for each traffic category
label_counts = df[label_col].value_counts()

# Display traffic label counts
print("Traffic Label Distribution:\n")

print(label_counts)
```

```python
# Calculate the percentage of each traffic category
label_percentages = (
    df[label_col].value_counts(normalize=True) * 100
)

# Display percentages rounded to two decimal places
print("Traffic Label Percentages:\n")

print(label_percentages.round(2))
```

```python
# Create a summary table containing counts and percentages
label_summary = pd.DataFrame({
    "Count": df[label_col].value_counts(),
    "Percentage (%)": (
        df[label_col].value_counts(normalize=True) * 100
    )
})

# Display the summary table
display(label_summary.round(2))
```

```python
# Generate descriptive statistics for numerical columns
# This includes count, mean, standard deviation,
# minimum, maximum and quartiles.

numerical_summary = df.describe().T

# Display the statistical summary
display(numerical_summary)
```

```python
# Identify numerical columns
numerical_columns = df.select_dtypes(
    include="number"
).columns.tolist()

# Identify non-numerical columns
categorical_columns = df.select_dtypes(
    exclude="number"
).columns.tolist()

# Display numerical column information
print("Number of Numerical Columns:", len(numerical_columns))

print("\nNumerical Columns:")
print(numerical_columns)

# Display categorical column information
print("\nNumber of Categorical Columns:", len(categorical_columns))

print("\nCategorical Columns:")
print(categorical_columns)
```

```python
# Select only numerical columns
numerical_df = df.select_dtypes(include="number")

# Count positive infinity values
positive_inf = np.isposinf(numerical_df).sum().sum()

# Count negative infinity values
negative_inf = np.isneginf(numerical_df).sum().sum()

# Calculate total infinity values
total_inf = positive_inf + negative_inf

# Display results
print("Positive Infinity Values:", positive_inf)
print("Negative Infinity Values:", negative_inf)
print("Total Infinity Values:", total_inf)
```

```python
# Count infinity values in each numerical column
inf_counts = pd.Series(
    np.isinf(numerical_df).sum(),
    index=numerical_df.columns
)

# Display only columns containing infinity
print("Columns Containing Infinity:\n")

print(inf_counts[inf_counts > 0])
```

```python
# Identify rows containing infinity
inf_mask = np.isinf(
    df[["Flow Bytes/s", " Flow Packets/s"]]
).any(axis=1)

# Count affected rows
print("Rows Containing Infinity:", inf_mask.sum())

# Find the exact target column name
label_col = next(
    col for col in df.columns
    if col.strip() == "Label"
)

# Display affected records using exact column names
affected_rows = df.loc[
    inf_mask,
    [
        " Flow Duration",
        "Flow Bytes/s",
        " Flow Packets/s",
        label_col
    ]
]

# Display first ten affected records
display(affected_rows.head(10))
```

```python
# Count missing values in every column
missing_counts = df.isnull().sum()

# Calculate total missing values
total_missing = missing_counts.sum()

# Display total missing values
print("Total Missing Values:", total_missing)

# Display only columns containing missing values
print("\nColumns Containing Missing Values:\n")

print(missing_counts[missing_counts > 0])
```

```python
# Identify rows where Flow Bytes/s is missing
missing_mask = df["Flow Bytes/s"].isna()

# Count affected rows
print("Rows With Missing Flow Bytes/s:", missing_mask.sum())

# Find the exact target column name
label_col = next(
    col for col in df.columns
    if col.strip() == "Label"
)

# Display selected columns from affected records
missing_rows = df.loc[
    missing_mask,
    [
        " Flow Duration",
        "Total Length of Fwd Packets",
        " Total Length of Bwd Packets",
        "Flow Bytes/s",
        label_col
    ]
]

# Display the first ten affected records
display(missing_rows.head(10))
```

```python
# Calculate memory usage of each column
memory_usage = df.memory_usage(deep=True)

# Calculate total memory usage in bytes
total_memory_bytes = memory_usage.sum()

# Convert bytes into megabytes
total_memory_mb = total_memory_bytes / (1024 ** 2)

# Display memory usage
print("Total Memory Usage:", round(total_memory_mb, 2), "MB")
```
