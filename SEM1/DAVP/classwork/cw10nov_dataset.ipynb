```python
import pandas as pd

df = pd.read_excel('/home/ayushhh/CUTM/SEM1/DAVP/student_academic_performance_big_dataset.xlsx')

print(df.head(5))
```

```python
# Qu 1
print(df[['Name', 'Marks']])
```

```python
#qu 2
high_performers = df[df['Marks'] >= 80]

print('Students with 80 and above marks:')
print(high_performers[['Name', 'Marks']])
```

```python
# Qn 3
# Filter students with attendance percentage < 75
low_attendance = df[df['Attendance_Percent'] < 75]

print('Students with less than 75% attendance:')
print(low_attendance[['Name', 'Attendance_Percent']])
```

```python
# Qn 4
import matplotlib.pyplot as plt

# Calculate average marks per department
dept_avg = df.groupby('Department')['Marks'].mean()
print('Average Marks by Department:')
print(dept_avg)

# Plotting the bar graph
dept_avg.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Marks by Department')
plt.xlabel('Department')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Define grading criteria function
def assign_grade(marks):
  if marks >= 90:
    return 'A+'
  elif marks >= 80:
    return 'A'
  elif marks >= 70:
    return 'B'
  elif marks >= 60:
    return 'C'
  elif marks >= 50:
    return 'D'
  else:
    return 'F'


# Apply grading to create a new 'Grade' column
df['Grade'] = df['Marks'].apply(assign_grade)

# Display name, marks, and grade
print(df[['Name', 'Marks', 'Grade']].head(10))
```

```python
# Find highest and lowest values
highest_marks = df['Marks'].max()
lowest_marks = df['Marks'].min()

print(f'Highest Marks in Dataset: {highest_marks}')
print(f'Lowest Marks in Dataset: {lowest_marks}')

# To find the specific student details:
top_student = df.loc[df['Marks'].idxmax()]
bottom_student = df.loc[df['Marks'].idxmin()]

print('\nTop Student Details:')
print(top_student[['Name', 'Department', 'Marks']])

print('\nLowest Student Details:')
print(bottom_student[['Name', 'Department', 'Marks']])
```

```python
# Count students in each department
dept_counts = df['Department'].value_counts()
print('Total Students per Department:')
print(dept_counts)

# Plotting the bar chart
dept_counts.plot(kind='bar', color='orange', edgecolor='black')
plt.title('Total Number of Students from Each Department')
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
