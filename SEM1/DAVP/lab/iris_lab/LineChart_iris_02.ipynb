```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

avg = df.groupby('Species')['SepalLengthCm'].mean()

plt.figure(figsize=(9, 5))

plt.plot(
    avg.index,
    avg.values,
    color='purple',
    linestyle='--',
    linewidth=2.5,
    marker='o',
    markersize=10,
    markerfacecolor='yellow',
    markeredgecolor='black',
    label='Average Sepal Length'
)

plt.title("Average Sepal Length by Iris Species",
          fontsize=16, fontweight='bold')

plt.xlabel("Species")
plt.ylabel("Average Sepal Length (cm)")

plt.legend(loc='best')
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
