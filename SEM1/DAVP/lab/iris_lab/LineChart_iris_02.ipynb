```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

plt.plot(df.SepalLengthCm, df.SepalWidthCm,
         marker="o", color="red",
         linestyle="--", markersize=4,
         alpha=0.7, label="Iris Data")

plt.title("Advanced Line Chart")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()
plt.grid()
plt.show()
```
