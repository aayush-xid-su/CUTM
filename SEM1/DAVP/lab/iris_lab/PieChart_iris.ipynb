```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

species_count = df["Species"].value_counts()

plt.pie(
    species_count,
    labels=species_count.index,
    autopct="%1.1f%%"
)

plt.title("Iris Species Distribution")

plt.show()
```
