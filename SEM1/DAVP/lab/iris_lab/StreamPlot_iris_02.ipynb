```python
import matplotlib.pyplot as plt
import numpy as np

x, y = np.meshgrid(np.linspace(0, 5, 20),np.linspace(0, 3, 20))

u = -y
v = x

plt.streamplot(x, y, u, v,density=1.5, color="red",linewidth=1.5, arrowsize=2)

plt.title("Advanced Streamplot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid()
plt.show()
```
