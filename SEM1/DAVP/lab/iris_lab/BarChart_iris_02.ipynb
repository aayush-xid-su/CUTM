```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

avg = df.groupby('Species')['PetalLengthCm'].mean()

plt.figure(figsize=(9, 5))

bars = plt.bar(
    avg.index,
    avg.values,
    color=['red', 'green', 'blue'],
    edgecolor='black',
    linewidth=1.5,
    width=0.5
)

plt.bar_label(bars, fmt='%.2f', padding=4)

plt.title("Average Petal Length by Iris Species",
          fontsize=16, fontweight='bold')

plt.xlabel("Species", fontsize=12)
plt.ylabel("Average Petal Length (cm)", fontsize=12)

plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
