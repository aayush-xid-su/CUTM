```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")
x = df.Species.value_counts()

plt.pie(x, labels=x.index,
        autopct="%.1f%%",
        colors=["red", "green", "blue"],
        explode=[0.1, 0, 0],
        startangle=90, shadow=True)

plt.title("Advanced Pie Chart")
plt.legend()
plt.show()
```
