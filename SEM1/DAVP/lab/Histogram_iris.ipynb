```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

plt.hist(df["PetalLengthCm"],
         bins=15,
         color='black',
         edgecolor='black')

plt.title("Petal Length Distribution")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")

plt.grid(axis='y')

plt.show()
```
