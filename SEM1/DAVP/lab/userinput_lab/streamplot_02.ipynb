```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)

X, Y = np.meshgrid(x, y)

U = -Y
V = X

speed = np.sqrt(U**2 + V**2)

plt.figure(figsize=(6, 4))

plt.streamplot(X, Y, U, V,color=speed,cmap='plasma',linewidth=1.5, density=1.8,arrowsize=1.5,arrowstyle='->',minlength=0.1)

plt.colorbar(label='Flow Speed')

plt.title("Customized Streamplot",fontsize=16, fontweight='bold')

plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.grid(True, alpha=0.3)
plt.axis('equal')

plt.tight_layout()
plt.show()
```
