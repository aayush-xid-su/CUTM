```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [10, 20, 15, 30, 25, 40, 35, 50]

plt.stem(x, y, linefmt="green", markerfmt="ro", basefmt="k-")

plt.title("Advanced Stem Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.grid(axis="y", linestyle="--")
plt.legend(["Data Points"])
plt.show()
```
