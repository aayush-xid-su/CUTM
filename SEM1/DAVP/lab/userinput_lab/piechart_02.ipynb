```python
import matplotlib.pyplot as plt

languages = ['Python', 'Java', 'C', 'C++']
students = [40, 25, 20, 15]

colors = ['red', 'blue', 'green', 'orange']

plt.figure(figsize=(3, 3))

plt.pie(students,labels=languages,colors=colors,autopct='%1.1f%%',startangle=70,explode=(0.1, 0, 0, 0),
    shadow=True,counterclock=False,wedgeprops={'edgecolor': 'black', 'linewidth': 1.5},
    textprops={'fontsize': 11}
)

plt.title("Programming Language Preferences",fontsize=16, fontweight='bold')

plt.legend(loc='lower right', title='Languages')
plt.axis('equal')

plt.tight_layout()
plt.show()
```
