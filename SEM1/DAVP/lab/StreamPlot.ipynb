```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

x = df["PetalLengthCm"]
y = df["PetalWidthCm"]

H, xedges, yedges = np.histogram2d(x, y, bins=20)

xc = (xedges[:-1] + xedges[1:]) / 2
yc = (yedges[:-1] + yedges[1:]) / 2

X, Y = np.meshgrid(xc, yc)

grad_y, grad_x = np.gradient(H.T, yc, xc)

plt.streamplot(X, Y, grad_x, grad_y,
               density=1.5,
               color='blue')

plt.title("Streamplot of Iris Petal Measurements")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.show()
```
