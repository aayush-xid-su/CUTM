```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

count = df["Species"].value_counts()

plt.pie(count, labels=count.index,utopct="%1.1f%%", startangle=90,explode=[0.05, 0.05, 0.05],shadow=True)

plt.title("Advanced Iris Pie Chart")
plt.legend(title="Species")
plt.show()
```
