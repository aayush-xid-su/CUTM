```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

plt.scatter(df["SepalLengthCm"], df["SepalWidthCm"],c=df["PetalLengthCm"], cmap="viridis",s=60, alpha=0.7, edgecolor="black")

plt.colorbar(label="Petal Length")
plt.title("Advanced Iris Scatter Chart")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.grid(True)
plt.show()
```
