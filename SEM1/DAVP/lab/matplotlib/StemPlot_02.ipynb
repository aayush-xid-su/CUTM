```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/ayushhh/CUTM/SEM1/DAVP/IRIS.csv")

plt.stem(df["Id"][:20], df["SepalLengthCm"][:20],
         linefmt="green", markerfmt="ro",
         basefmt="k-", label="Sepal Length")

plt.title("Advanced Iris Stem Plot")
plt.xlabel("Id")
plt.ylabel("Sepal Length")
plt.grid(axis="y", linestyle="--")
plt.legend()
plt.show()
```
