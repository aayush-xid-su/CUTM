```python
import numpy as np
print('1 D array')
a = np.array([1,2,3])
print(a)
```

```python
print('2D Array')
b = np.array([[1,2,3],[4,5,6]])
print(b)
```

```python
# tell what type of data
type(a)
type(b) 
```

```python
# size in array
print(a.size)
print(b.size)
```

```python
# dtype tell what type of data 
print(a.dtype)
print(b.dtype)
```
