| Concept | Syntax | Purpose |

| Select one column | `df["Species"]` | Returns a Series |
| Select multiple columns | `df[["Species", "PetalLengthCm"]]` | Returns a DataFrame |
| Select a row by position | `df.iloc[0]` | Selects the first row |
| Select by index label | `df.loc[0]` | Selects the row labelled `0` |
| Select a value by position | `df.iloc[0, 5]` | Selects a value using integer positions |
| Select a value by label | `df.loc[0, "Species"]` | Selects a value using labels |
| Filter rows | `df[df["PetalLengthCm"] > 5]` | Returns rows satisfying a condition |
| Combine with AND | `condition1 & condition2` | Both conditions must be true |
| Combine with OR | `condition1 \| condition2` | At least one condition must be true |
| Negate a condition | `~condition` | Inverts a Boolean mask |
| Filter a range | `df["SepalLengthCm"].between(5, 6)` | Selects values within a range |
| Filter multiple values | `df["Species"].isin([...])` | Selects specified categories |
| Sort ascending | `df.sort_values(by="PetalLengthCm")` | Smallest to largest |
| Sort descending | `df.sort_values(by="PetalLengthCm", ascending=False)` | Largest to smallest |
| Display first 10 rows | `df.head(10)` | Shows the first 10 rows |
| Count matching rows | `df[condition].shape[0]` | Returns the number of matches |
