```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
roll = [1,2,3,4,5]
marks = [11,22,33,44,55]

plt.plot(roll,marks)
plt.show()
```

```python
plt.plot(roll, marks, linestyle='--',color='r')
```

```python
plt.plot(roll, marks, linestyle='--',linewidth=5)
```

```python
mcaBroll = [1,2,3,4,5,6,7,8,9]
mcaBmark = [11,22,33,44,55,66,77,88,99]

plt.Figure(figsize=(8,8))
plt.xticks(np.arange(0,10,1))
plt.yticks(np.arange(11,100,10))

plt.plot(mcaBroll, mcaBmark, 'r')

plt.xlabel('Roll')
plt.ylabel('Marks')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.xticks(np.arange(1,10,1))
plt.yticks(np.arange(11,110,10))

plt.plot(mcaBroll, 'r-')
plt.plot(mcaBmark, 'g*')

plt.xlabel('Roll')
plt.ylabel('Marks')
plt.show()
```
