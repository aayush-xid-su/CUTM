```python
import matplotlib.pyplot as plt
x = ["Java","Python","c++","HTML","DS"]
y = [40,50,60,70,80]
c = ["b","g","y","m","r"]
plt.bar(x,y,color=c,linewidth = 5, linestyle =":",edgecolor="black")
plt.xlabel("Book",color="g")
plt.ylabel("price",color="y")
plt.title("sales",color="m")
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40,50]
plt.pie(x)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40,50]
y = ["java","python","DS","c++","HTML"]
plt.pie(x,labels=y)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40,50]
y = ["Java","Python","HTML","C++","DS"]
c = ["r","g","b","y","m"]
ex = [0.4,0.0,0.0,0.0,0.0]
plt.pie(x,labels=y,explode=ex,colors=c,autopct="%0.f%%",shadow=True)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40,50]
y = ["Java","Python","HTML","C++","DS"]
c = ["r","g","b","y","m"]
ex = [0.4,0.2,0.0,0.0,0.0]
plt.pie(x,labels=y,explode=ex,colors=c,autopct="%0.f%%",shadow=True,startangle=90,labeldistance=1.5,textprops={"fontsize":20})
plt.show()
```
