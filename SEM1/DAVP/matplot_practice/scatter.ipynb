```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
roll = [1,2,3,4,5,6,7,8,9]
marks = [88,99,67,45,45,68,24,96,84]

plt.scatter(roll, marks)
plt.show()
```

```python
plt.scatter(roll, marks, color="green")
plt.show()
```

```python
plt.scatter(roll, marks, marker="*")
plt.show()
```

```python
#using plot how scatter graph works
plt.plot(roll, marks)
plt.show()
```

```python
# 'g' color and o for symbol and markersize for size 
plt.plot(roll, marks, 'go', markersize=15)
plt.show()
```

```python
# plotting roll amd marks using x,yticks and x,ylabel 
MCArollA = [1,2,3,4,5]
MCAmarksA = [11,24,35,47,59]

plt.xticks(np.arange(0,10,1))
plt.yticks(np.arange(0,70,10))

plt.plot(MCArollA, MCAmarksA, 'go', markersize=10)
plt.xlabel('ROLL NO')
plt.ylabel('MARKS')
plt.show()
```

```python
# combine both class roll and marks
MCArollA = [1,2,3,4,5]
MCAmarksA = [11,24,35,47,59]

MCArollB = [10,11,12,13,14]
MCAmarksB = [79,76,82,85,88]

plt.xticks(np.arange(0,100,1))
plt.yticks(np.arange(0,100,10))

plt.plot(MCArollA, MCAmarksA, 'ro', markersize=10)
plt.plot(MCArollB, MCAmarksB, 'g*', markersize=10)

plt.xlabel('ROLL NO')
plt.ylabel('MARKS')
plt.show()
```
