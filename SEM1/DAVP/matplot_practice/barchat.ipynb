```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
sub = ['c','python','uiux','dbms','swe']
marks = [99,88,96,94,93]

plt.bar(sub,marks)
plt.show()
```

```python
# color used to give color 
plt.bar(sub,marks,color='r')
plt.show()
```

```python
# different colors 
colors = ['red','green','blue','orange','purple']
plt.bar(sub,marks,color=colors)
plt.show()
```

```python
# width use to thickness of bar
colors = ['red','green','blue','orange','purple']
plt.bar(sub,marks,color=colors, width = 0.2)
plt.show()
```

```python
# edgecolor used to color the edge of the bar
plt.bar(sub,marks,color='red', edgecolor = 'green')
plt.show()
```

```python
# linewidth use to inc dec the thickness of edge of bar
plt.bar(sub,marks,color='red', edgecolor = 'green', linewidth=4)
plt.show()
```

```python
# linestyle use to styple the edge of bar
plt.bar(sub,marks,color='red', edgecolor = 'green', linewidth=5,linestyle='--')
plt.show()
```

```python
# barh for horizontal bar cart
colors = ['red','green','blue','orange','purple']
plt.barh(sub,marks,color=colors)
plt.show()
```

```python
mcasub = ['c','python','uiux','dbms','swe']
marksA = [99,98,96,94,93]
marksB = [87,89,84,86,81]

plt.figure(figsize=(4,4))

plt.bar(mcasub,marksA,color='r')
plt.bar(mcasub,marksB,color='g')

plt.xlabel('sub')
plt.ylabel('Marks')
plt.show()
```

```python
mcasub = ['c','python','uiux','dbms','swe']
marksA = [99,98,96,94,93]
marksB = [87,89,84,86,81]

plt.figure(figsize=(6,5))

sublen = np.arange(len(mcasub))
width = 0.4

plt.bar(sublen, marksA, width=width, color='r')
plt.bar(sublen + width, marksB, width=width, color='g')

plt.xlabel('sub')
plt.ylabel('Marks')
plt.show()
```

```python
mcasub = ['c','python','uiux','dbms','swe']
marksA = [99,98,96,94,93]
marksB = [87,89,84,86,81]

plt.figure(figsize=(6,5))

sublen = np.arange(len(mcasub))
width = 0.4

plt.bar(sublen, marksA, width=width, color='r')
plt.bar(sublen + width, marksB, width=width, color='r', alpha=0.5)

plt.xlabel('sub')
plt.ylabel('Marks')
plt.show()
```
